## 1. Definição da Problemática
### Contexto e Relevância:

O hipertireoidismo é uma condição clínica caracterizada pela produção excessiva de hormônios tireoidianos, podendo causar sintomas como ansiedade, taquicardia, perda de peso, fadiga, alterações metabólicas e complicações cardiovasculares graves.

A identificação precoce da doença é fundamental para reduzir riscos clínicos e iniciar o tratamento adequado o mais rápido possível. Entretanto, o diagnóstico pode ser complexo devido à variabilidade dos sintomas, à presença de doenças associadas e à possibilidade de resultados laboratoriais pouco conclusivos.

Nesse contexto, modelos preditivos baseados em machine learning podem auxiliar profissionais da saúde a:

* Identificar pacientes com maior probabilidade de hipertireoidismo
* Detectar padrões relevantes em exames laboratoriais e históricos clínicos
* Auxiliar diagnósticos em casos inconclusivos
* Apoiar decisões médicas baseadas em dados
* Reduzir falhas diagnósticas
* Priorizar pacientes com maior risco clínico

Além disso, minimizar falsos negativos é um fator crítico neste cenário, uma vez que pacientes doentes não diagnosticados podem evoluir para quadros mais graves e complicações sistêmicas importantes.

### Problema Específico a Resolver:

**"Como desenvolver um modelo preditivo capaz de diagnosticar hipertireoidismo com alta sensibilidade, minimizando falsos negativos e identificando padrões clínicos relevantes a partir de dados laboratoriais e históricos médicos, especialmente em casos onde os exames apresentam resultados pouco conclusivos?"**

### Objetivo Geral:

Desenvolver um modelo de machine learning para diagnóstico de hipertireoidismo que permita:

1. Identificar pacientes com alta probabilidade de hipertireoidismo
2. Maximizar a identificação de verdadeiros positivos
3. Minimizar falsos negativos, reduzindo o risco de pacientes doentes não serem diagnosticados
4. Detectar padrões relevantes entre exames hormonais e condições clínicas
5. Auxiliar profissionais da saúde em diagnósticos complexos
6. Avaliar quais variáveis possuem maior influência na predição da doença
7. Construir uma análise exploratória robusta para compreensão dos dados clínicos e laboratoriais

### Diferencial do Projeto:

Este projeto posiciona-se na intersecção de:

* **Ciência de Dados Aplicada à Saúde:** utilização de machine learning para apoio diagnóstico em dados médicos
* **Diagnóstico Assistido por IA:** suporte à tomada de decisão clínica baseada em padrões laboratoriais e históricos médicos
* **Análise Exploratória de Dados Médicos:** identificação de padrões relevantes entre hormônios tireoidianos, tratamentos e condições associadas
* **Machine Learning Sensível a Casos Críticos:** foco na minimização de falsos negativos e maximização da detecção de pacientes realmente doentes
* **Analytics Preditivo em Saúde:** utilização de modelos preditivos para aumentar precisão diagnóstica em cenários clínicos complexos

### Base de Dados

O dataset utilizado contém informações clínicas, laboratoriais e históricas de pacientes avaliados para distúrbios da tireoide. A variável alvo do projeto corresponde ao diagnóstico binário de hipertireoidismo, permitindo classificar pacientes entre casos positivos e negativos para a doença.

#### Descrição das variáveis:

| Variável | Descrição |
|---|---|
| `age` | Indica a idade do paciente. |
| `sex` | Indica o gênero sexual do paciente |
| `on thyroxine` | Indica se o paciente está fazendo uso de tiroxina (hormônio tireoidiano sintético utilizado no tratamento de hipotireoidismo). |
| `query on thyroxine` | Indica suspeita ou investigação clínica relacionada ao uso ou necessidade de tiroxina. |
| `on antithyroid medication` | Indica se o paciente utiliza medicação antitireoidiana para controle do hipertireoidismo. |
| `sick` | Indica se o paciente apresenta alguma condição de doença no momento da coleta dos dados. |
| `I131 treatment` | Indica se o paciente realizou tratamento com Iodo Radioativo (Iodo-131), geralmente utilizado em hipertireoidismo ou câncer de tireoide. |
| `query hypothyroid` | Indica suspeita clínica de hipotireoidismo. |
| `query hyperthyroid` | Indica suspeita clínica de hipertireoidismo. |
| `lithium` | Indica se o paciente faz uso de lítio, medicamento que pode afetar a função tireoidiana. |
| `goitre` | Indica presença de bócio (aumento anormal da glândula tireoide). |
| `tumor` | Indica presença ou suspeita de tumor tireoidiano. |
| `hypopituitary` | Indica presença de hipopituitarismo, condição em que a hipófise produz hormônios em quantidade insuficiente. |
| `psych` | Indica presença de condição psiquiátrica ou uso de tratamento psiquiátrico. |
| `TSH measured` | Indica se o exame de TSH foi realizado. |
| `TSH` | Valor do hormônio TSH (Thyroid Stimulating Hormone), principal marcador da função tireoidiana. |
| `T3 measured` | Indica se o exame de T3 foi realizado. |
| `T3` | Valor do hormônio T3 (Triiodotironina). |
| `TT4 measured` | Indica se o exame de TT4 foi realizado. |
| `TT4` | Valor de T4 Total (Total Thyroxine). |
| `T4U measured` | Indica se o exame de T4 Uptake foi realizado. |
| `T4U` | Resultado do exame T4 Uptake, utilizado para avaliar proteínas transportadoras dos hormônios tireoidianos. |
| `FTI measured` | Indica se o exame FTI foi realizado. |
| `FTI` | Free Thyroxine Index, índice utilizado para estimar a quantidade de T4 livre circulante. |
| `TBG measured` | Indica se o exame TBG foi realizado. |
| `TBG` | Valor da proteína TBG (Thyroxine Binding Globulin), responsável pelo transporte dos hormônios tireoidianos no sangue. |
| `referral source` | Fonte de encaminhamento do paciente (hospital, laboratório, clínica, médico etc.). |
| `binaryClass` | Classe alvo binária do dataset, geralmente indicando presença ou ausência de doença tireoidiana. |
